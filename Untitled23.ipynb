### 3.1. Overview of Consolidated Bank Data (`bank_df`)

```python
import pandas as pd

# Ensure all necessary dataframes are loaded and processed in one place for robustness
# Reload initial dataframes if the kernel state was lost
pnl_df = pd.read_excel('/content/CoffeeShop_PnL_Jun2023.xlsx')
pos_df = pd.read_excel('/content/CoffeeShop_POS_Jun2023.xlsx')
bank_data_day1_df = pd.read_excel('/content/CoffeeShop_BankData_Day1_Jun2023.xlsx')
bank_data_day2_df = pd.read_excel('/content/CoffeeShop_BankData_Day2_Jun2023.xlsx')
commentary_df = pd.read_excel('/content/CoffeeShop_Commentary_Jun2023.xlsx')

# Process pos_df: convert 'Order Time' column
pos_df['Order Time'] = pd.to_datetime(pos_df['transaction_date'].astype(str) + ' ' + pos_df['transaction_time'].astype(str))

# Consolidate bank data into bank_df
bank_df = pd.concat([bank_data_day1_df, bank_data_day2_df], ignore_index=True)

print("Dataframes 'pos_df' and 'bank_df' are now prepared.")
```

### 3.2. Overview of Point of Sale Data (`pos_df`)

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Descriptive statistics for transaction_qty and product_price in pos_df
print("Descriptive statistics for 'transaction_qty' in pos_df:")
display(pos_df['transaction_qty'].describe())

# Let's look at the top 10 most sold items
top_items = pos_df.groupby('product_detail')['transaction_qty'].sum().nlargest(10)

plt.figure(figsize=(12, 6))
sns.barplot(x=top_items.index, y=top_items.values, palette='viridis')
plt.title('Top 10 Most Sold Items')
plt.xlabel('Item')
plt.ylabel('Total Quantity Sold')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()

# Distribution of payment types
plt.figure(figsize=(8, 6))
sns.countplot(x='Payment Type', data=pos_df, palette='pastel')
plt.title('Distribution of Payment Types')
plt.xlabel('Payment Type')
plt.ylabel('Count')
plt.show()
```

### 4.1. Financial Performance Overview (PnL Data)

Let's first review the Profit & Loss statement to understand the overall financial health and identify key trends in revenues and expenses.

```python
# Assuming 'pnl_df' is already loaded and cleaned from earlier steps

# Filter for June 2023 actuals (or latest available month if June is not complete)
# Assuming 'Month' column is string month name, and 'Year' is integer
pnl_june_actuals = pnl_df[(pnl_df['Month'] == 'Jun') & (pnl_df['Year'] == 2023) & (pnl_df['Version'] == 'Actuals')]

# Group by Account to summarize revenues and expenses
financial_summary = pnl_june_actuals.groupby('Account')['Value'].sum().reset_index()
display(financial_summary)

# Calculate Net Profit/Loss if possible (requires categorizing Accounts into Revenues and Expenses)
revenue = financial_summary[financial_summary['Account'].str.contains('Revenue')]['Value'].sum()
expenses = financial_summary[~financial_summary['Account'].str.contains('Revenue')]['Value'].sum()

net_profit = revenue - expenses
print(f"\nEstimated Net Profit/Loss for June 2023 (Actuals): ${net_profit:,.2f}")

# Visualize PnL components
plt.figure(figsize=(12, 7))
sns.barplot(x='Value', y='Account', data=financial_summary, palette='coolwarm')
plt.title('PnL Summary for June 2023 (Actuals)')
plt.xlabel('Value ($)')
plt.ylabel('Account')
plt.grid(axis='x', alpha=0.75)
plt.tight_layout()
plt.show()
```

### 4.2. Sales Performance Insights (POS Data)

Moving to the Point of Sale (POS) data, we can delve into sales patterns, popular products, and customer behavior.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Ensure 'pos_df' is loaded and 'Order Time' is converted to datetime
pos_df = pd.read_excel('/content/CoffeeShop_POS_Jun2023.xlsx')
pos_df['Order Time'] = pd.to_datetime(pos_df['transaction_date'].astype(str) + ' ' + pos_df['transaction_time'].astype(str))

# Sales trends over time (e.g., daily sales)
pos_df['Order Date'] = pos_df['Order Time'].dt.date
daily_sales = pos_df.groupby('Order Date')['transaction_qty'].sum().reset_index()

plt.figure(figsize=(14, 7))
sns.lineplot(x='Order Date', y='transaction_qty', data=daily_sales)
plt.title('Daily Transaction Quantity Trend')
plt.xlabel('Date')
plt.ylabel('Total Transaction Quantity')
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()

# Average transaction value (requires product_price or similar, assuming here transaction_qty as a proxy for value if no price column is directly available yet)
# If 'product_price' is available, this calculation would be more accurate
pos_df['transaction_value'] = pos_df['transaction_qty'] # Placeholder, ideal would be transaction_qty * product_price

average_transaction_value = pos_df['transaction_value'].mean()
print(f"\nAverage Transaction Value (based on quantity): {average_transaction_value:.2f}")
```
