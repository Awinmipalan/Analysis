```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

def robust_read_file(path):
    """Attempts to read a file as Excel, falling back to CSV with deeper inspection."""
    try:
        return pd.read_excel(path)
    except Exception as e:
        print(f"⚠️ {os.path.basename(path)} is not a valid .xlsx. Checking file head...")
        # Peek at the file to diagnose
        with open(path, 'r', encoding='utf-8', errors='replace') as f:
            head = [f.readline() for _ in range(3)]
            print(f"File head sample: {head}")

        # Try parsing with most common CSV fallbacks
        for enc in ['utf-8', 'cp1252', 'latin1']:
            for sep in [',', '\t', ';', '|']:
                try:
                    df = pd.read_csv(path, encoding=enc, sep=sep, on_bad_lines='skip')
                    if df.shape[1] > 1:
                        print(f"Successfully parsed with encoding={enc} and sep='{sep}'")
                        return df
                except:
                    continue
        raise Exception(f"Could not parse {path}. Inspection suggests check needed.")

required_files = [
    '/content/CoffeeShop_PnL_Jun2023.xlsx',
    '/content/CoffeeShop_POS_Jun2023.xlsx',
    '/content/CoffeeShop_BankData_Day1_Jun2023.xlsx',
    '/content/CoffeeShop_BankData_Day2_Jun2023.xlsx',
    '/content/CoffeeShop_Commentary_Jun2023.xlsx'
]

missing_files = [f for f in required_files if not os.path.exists(f)]

if missing_files:
    print("❌ WORKFLOW BLOCKED: Missing Data Files")
else:
    try:
        pnl_df = robust_read_file('/content/CoffeeShop_PnL_Jun2023.xlsx')
        pos_df = robust_read_file('/content/CoffeeShop_POS_Jun2023.xlsx')
        bank_d1 = robust_read_file('/content/CoffeeShop_BankData_Day1_Jun2023.xlsx')
        bank_d2 = robust_read_file('/content/CoffeeShop_BankData_Day2_Jun2023.xlsx')
        commentary_df = robust_read_file('/content/CoffeeShop_Commentary_Jun2023.xlsx')

        # Standardize for cross-analysis
        bank_df = pd.concat([bank_d1, bank_d2], ignore_index=True)

        # Identify date column in POS (checking for common POS variations)
        date_col = next((c for c in pos_df.columns if 'date' in c.lower()), None)
        time_col = next((c for c in pos_df.columns if 'time' in c.lower()), None)

        if date_col and time_col:
            pos_df['Order Time'] = pd.to_datetime(pos_df[date_col].astype(str) + ' ' + pos_df[time_col].astype(str))
            pos_df['Order Date'] = pos_df['Order Time'].dt.date

        # Metrics
        june_actuals = pnl_df[(pnl_df['Month'] == 'Jun') & (pnl_df['Year'] == 2023)]
        revenue = june_actuals[june_actuals['Account'].str.contains('Revenue', case=False)]['Value'].sum()
        expenses = june_actuals[~june_actuals['Account'].str.contains('Revenue', case=False)]['Value'].sum()
        net_profit = revenue - expenses

        print('✅ WORKFLOW READY: Data successfully integrated.')
    except Exception as final_error:
        print(f"❌ CRITICAL ERROR: {final_error}")
```

### 5.2. How do our customers pay and when do they visit?
*Understanding customer convenience and peak times.*

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))

# Payment Methods
# Using 'Payment Type' or fallback to identifying payment column
payment_col = 'Payment Type' if 'Payment Type' in pos_df.columns else pos_df.columns[-1]
payment_counts = pos_df[payment_col].value_counts()
ax1.pie(payment_counts, labels=payment_counts.index, autopct='%1.1f%%', colors=sns.color_palette('pastel'))
ax1.set_title('How Customers Pay Us', fontsize=14, fontweight='bold')

# Weekly Sales Pulse
pos_df['Order Date'] = pd.to_datetime(pos_df['transaction_date']).dt.date
daily_sales = pos_df.groupby('Order Date')['transaction_qty'].sum().reset_index()
daily_sales['Day'] = pd.to_datetime(daily_sales['Order Date']).dt.day_name()
day_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
weekly_pulse = daily_sales.groupby('Day')['transaction_qty'].mean().reindex(day_order)

ax2.plot(weekly_pulse.index, weekly_pulse.values, marker='o', linestyle='-', color='teal', linewidth=3)
ax2.set_title('Our Weekly "Heartbeat": Average Sales by Day', fontsize=14, fontweight='bold')
ax2.set_ylabel('Average Units Sold')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

**Interpretation for Stakeholders:**
*   **Payments:** We see a mix of payment types. If digital payments are high, we must ensure our Wi-Fi and POS systems are robust. If cash is high, security and frequent bank runs are priorities.
*   **The Weekly Pulse:** The line chart shows our busiest days. We should align our staff schedules to these peaks to ensure customers aren't waiting too long for their coffee.
